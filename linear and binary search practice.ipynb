```python
#linear search
def linear_search(arr, target):
    for i in range (len(arr)):
        if arr[i] == target:
            return i
    return -1
numbers = [12, 8, 25, 40, 18]
target = 18
result = linear_search(numbers, target)
if result != -1:
    print("linear_search : Elements found at index", result)
else:
    print("linear_search : Elements not found")
```

```python
#linear search   -   student roll number
def linear_search(roll_number, target):
    for i in range (len(roll_number)):
        if roll_number[i] == target:
            return i
    return -1
roll_numbers = [101, 102, 103, 105, 109]
target = 103
result = linear_search(roll_numbers, target)
if result != -1:
    print("linear_search : Elements not found at index", result)
else:
    print("linear_search : Elements not found")
```

```python
#linear search - product 
def linear_search(products, target):
    for i in range (len(products)):
        if products[i] == target:
            return i
    return -1
products = "mouse", "keyboard", "book", "laptop", "table"
target = "laptop"
result = linear_search(products, target)
if result != -1:
    print("linear_search : Elements not found at index", result)
else:
    print("linear_search : Elements not found")
```
