```python
def fun(x):
    if x == 0:
        return 0
    x = x - 1
    fun(x)
    print(x)
fun(3)
```

```python
# Print 1 to n using recursion
def fun(x):
    if x == 0:
        return 0
    fun(x - 1)
    print(x)
print(fun(6))
```

```python
# print 1 to n using recrusion(reverse)
def fun(x):
    if x == 0:
        return 0
    print(x)
    fun(x - 1)
    
fun(6)
```

```python
def factorial(x):
    if x == 0 or x == 1:
        return 1
    return x * factorial(x - 1)
print(factorial(5))
```

```python
# Sum of digits in a number
# n = 456 o/p: 15
def sumofdigits(n):
    if n == 0:
        return 0
    return n % 10 + sumofdigits(n//10)
print(sumofdigits(456))
print(sumofdigits(15346))
```

```python
# Find the square of a number using Lambda
s = lambda x:x**2
print(s(10))
```

```python
# Find the squares of list of integers using map and lambda
arr = [1,2,3,4,5]
print(list(map(lambda x: x**2 , arr)))
```

```python
# Check whether the number is even or odd using lambda
even_or_odd = lambda x:"even" if x%2 ==0 else "odd"
print(even_or_odd(10))
print(even_or_odd(9))
```

```python
# Find the frequency of each character
string = input()
count = {}
for char in string:
    if char in count:
        count[char] += 1
    else:
        count[char] = 1
print(count)
```

```python
# Find the frequency of each character
string = input()
count = {}
for char in string:
        count[char] = count.get(char, 0) + 1
print(count)
```
