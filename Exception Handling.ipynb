### Exception Handling : 
- It is one type for error, whenever an exception occurs it stops the program execution and it returns type of exception and followed by execution message.
### Types of Errors :
                1. Syntax error / Compiletime error    
                2. Runtime error 
                3. Logical error
`1. Syntax error / Compiletime error :`  <br>
                    -The Syntax error raise during the compilation time of the program.Those errors are call it as Syntax error.   <br>
                       eg: Intendation error,Syntax error.    <br> 
`2. Runtime error :`  <br>
                    - The Runtime errors raises during Runtime of the program.Those errors are called as Runtime error.   <br>
                       eg: KeyError, ValueError, Zero Division Error, IndexError, TypeError, etc...   <br>
                    - We can't handle syntax error but we can handle the runtime errors by using exception handling blocks.   <br>
                      (i.e. try-except)   <br>
`Exception Handling Blocks:`   <br>
`Syntax:`   <br>
       try:    <br>
          # block code that might contain exception.   <br>
       except:   <br>
          # exception handling block  <br>
       else:    <br>
          # this block executes when there is no exception raised in try block.   <br>
       finally:  <br>
          # this block always executs ever exception raised or not   <br>

```python
try:
    list = list[1,2,3]
    print(list[10])
    print("Hi")
except:
    print("Something wrong in try block")
else:
    print("No Exception raised in try block")
finally:
    print("Program Done")
```

```python
# try block with exception
try:
    list = list[1,2,3]
    print(list[2])
    print("Hi")
except:
    print("Something wrong in try block")
else:
    print("No Exception raised in try block")
finally:
    print("Program Done")
```

```python
# No exception in try block
# we cannot handle the syntax errors.
try:
    print("Hello')
except:
    print("Something wrong in try block")
else:
    print("No Exception raised in try block")
finally:
    print("Program Done")
```

```python
try:
    a = 10
    b = a.append(0)
    print(b)
except ZeroDivisionError as e:
    print(f"Zero division not possible: {e}")
except ValueError as e:
    print(f"Value Error raised in Try block: {e}")
    
```

```python
# It executes the specific exception block when specified exception raised otherwise, it executes the Common except block
try:
    a = 10
    b = 0
    c = a / b
    print(c)
except ZeroDivisionError as e:
    print(f"Zero division not possible: {e}")
except ValueError as e:
    print(f"Value Error raised in Try block: {e}")
except Exception as e: # common exception block with Exception Class
    print(f"Something went wrong in Try block: {e}")
```
