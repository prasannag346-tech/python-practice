```python
# Scopes of variable
# 1. Local scope
# 2. Enclosing scope 
# 3. Global scope
# 4. Builtin Scope



x = 10      # global scope variable
def outer():
    y = 20        # Enclosing scope variable
    def inner():
        z = 30        # LOcal scope variable
        print(z, y, x)  #print() - built in scope variable
    inner()
outer()
```

```python
# Using global scope variable in local scope variable
x = 100
def outer():
    y = x + 1
    print("Local Scope Variable value:", y)
outer()
print("Global Scope variable value:", x)
```

```python
# Updating the Global scope variable value in Local scope
# Cannot update global scope variable in local scope directly.
x = 100
def outer():
    x = x + 1
    print("Local scope variable value:", x)
outer()
print("Global Scope variable value:", x)
```

```python
# Updating the Global scope variable value in Local scope by using 'global' keyword
x = 100
def outer():
    global x
    x = x + 1
    print("Local scope variable value:", x)
outer()
print("Global Scope variable value:", x)
```

```python
# Updating the enclosing variable value in local scope variable
x = 10      # global scopr variable
def outer():
    y = 20        # Enclosing scope variable
    def inner():
        z = 30 
        y = y + z  # Local scope variable
        print(z, y, x)  #print() - built in scope variable
    inner()
outer()
    
```

```python
# Updating the enclosing variable value in local scope variable by using nonlocal
x = 10      # global scopr variable
def outer():
    y = 20        # Enclosing scope variable
    def inner():
        nonlocal y
        z = 30 
        y = y + z  # Local scope variable
        print(z, y, x)  #print() - Built in scope variable
    inner()
    print("Enclosing scope variable value:", y)
outer()
    
```
