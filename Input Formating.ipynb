### Input Formating

```python
# reading integer from user
num = int(input("Enter a number:"))
print(type(num))
print(num)
```

```python
# reading float value from user
val = float(input("Enter a number:"))
print(type(val))
print(val)
```

```python
#complex reading from user
val = complex(int(input())),int(input())
print(type(val))
print(val)
```

```python
# boolean reading as integer
val = bool(int(input()))
print(type(val))
print(val)
```

```python
# reading string value from user
s = input()
print(type(s))
print(s)
```

```python
# reading list of integers from user
l = list(input())
print(type(1))
print(l)
```

```python
# reading list of integers from user
lst = input().split()
input(lst)
```

```python
# list of integer from user
lst = list(map(int, input().split()))
print(lst)
```

```python
# reading list of strings from user
lst=input().split()
print(lst)


lst = list(map(str, input().split()))
print(lst)

```

```python
# reading tuple of integers from user
t1 = tuple(map(int,input().split()))
print(t1)
```
