### POLYMORPHISM : In Python "Polymorphysm", "poly" means "many" and "morphism" means "forms".
- In polymorphism, single function it shows different behaviours based on input context.
- In Python polymorphism can achieve through the following ways,  <br>
    `1. Through Operators`    <br>
    `2. Polymorphism through Functions`   <br>
    `3. Polymorphism through Methods`      <br>
- Types of Polymorphism through methods we have two types,   <br>
    ` 1. Run time polymorphism(method overriding)`    <br>
    ` 2. Complie time polymorphism(method overloading)`   <br>

`1. Polymorphism through Operators `: Here single operator, it shows different actions based on input context.   <br>
             Eg : '+' operator : It performs addition operation when inputs are integers as well as it forms string concatenation when inputs are strings.

    - 3+5 = 8                   
     code + gnan = codegnan
     eg for multiplication operator   
    - 2 * 3 = 6 
     2*code = "code" "code"   #repetation

`2. Polymorphism through Functions `: Here a function shows different actions based on input context.<br>
     eg :   <br>
            - len("hello")-->returns string length  <br>
            - len([1,2,3,4])-->returns list length

`3. Polymorphism Through Methods `  :    <br>
`        1. Method Overriding  `: (Through the inheritance)   <br>
                      - A Method defined in child class, that is exist in the parent class with the same name. Whenever if you call that method through the child class object, the child class method overrides the parent class method. This method is known as method overriding.  <br>
`        2. Method overloading  `: (During Compile Time)       <br>
                      - A class contains multiple methods with the same name with the different parameters list, In that case last defined method oberrides the perivous defined method.      <br>
                      - This is called as method overloading.   <br>
                      - Directly python doesn't supports the method overloading, but we can achieve same functionality throught default arguments, variable length positions, or variable length keyword arguments.


```python
# METHOD OVERRIDING
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self):
        return f"My name is {self.name} and i am {self.age} years old"
        

class Student(Person):
    def __init__(self, stdid, name, age, gender, branch):
        super().__init__(name, age, gender)
        self.stdid = stdid
        self.branch = branch
    def details(self):
        return f"My name is {self.name} and i am from {self.branch} Branch"   
s = Student(101, "Prasanna", 20, "female", "CSE")
print(s.details())
```

```python
# METHOD OVERRIDING
# Even if u define different number of parameters in methods, always child class method overrides parent class method.
class Person:
    def __init__(self, name, age, gender):
        self.name = name
        self.age = age
        self.gender = gender
    def details(self, location):
        return f"My name is {self.name} and i am staying in {location}"
        

class Student(Person):
    def __init__(self, stdid, name, age, gender, branch):
        super().__init__(name, age, gender)
        self.stdid = stdid
        self.branch = branch
    def details(self):
        print(super().details())
        return f"My name is {self.name} and i am from {self.branch} Branch"   
s = Student(101, "Prasanna", 20, "female", "CSE")
print(s.details())
print(s.details("India"))
```

```python
#METHOD OVERLOADING 
class Math:
    def add(self, a, b):
        return a+b
    def add(self, a, b, c):
        return a+b+c
    def add(self, a, b, c, d):
        return a+b+c+d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
#METHOD OVERLOADING 
# Directly python doesn't supports method overloading but, we can achive same functions through the default arguments or Arbitrary Arguments.
class Math:
    
    def add(self, a, b, c = 0, d = 0):
        return a+b+c+d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# __str__() magic method
class Point:
    def __init__(self,x):
        self.x = x
    def __str__(self):  # String magic method
        return f"This is point Class and Point value is {self.x}"
    def __add__(self, other):
        return Point(self.x + other.x)
    def __mul__(self, other):
        return Point(self.x * other.x)
    def __sub__(self, other):
        return Point(self.x - other.x)
    def __gt__(self, other):
        return Point(self.x > other.x)
    def __lt__(self, other):
        return Point(self.x < other.x)
    def __lshift__(self, other):
        return Point(self.x << other.x)
    def __rshift__(self, other):
        return Point(self.x >> other.x)

    
p1 = Point(5)
p2 = Point(3)
print(p1 + p2)
print(p1 * p2)
print(p1 - p2)
print(p1 > p2)
print(p1 < p2)
print(p1 << p2)
print(p1 >> p2)
```
