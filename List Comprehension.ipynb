```python
# Generate 1-n numbers using list comprehension
n = int(input())
res = [x for x in range(1, n+1)]
print(res)
```

```python
# First n even numbers using list comprehension
n = int(input())
res = [x * 2 for x in range(0, n)]
print(res)
```

```python
# Find the first n quibic numbers
n = int(input())
res = [x**3 for x in range(1, n+1)]
print(res)
```

```python
# Find the odd numbers between n1 to n2 
n1, n2 = map(int, input().split())
res = [x for x in range(n1, n2+1) if x%2 == 1]
print(res)  
```

```python
# in given arr of integers perform following operations
# if even - find square of a number
# if odd - find squareroot value(upto 2 decimal places)
arr = list(map(int, input().split()))
res = [x**2 if x%2==0 else round(x**0.5, 2) for x in arr]
print(res)        
```

```python
# 2 dices combination values
res = [(x, y) for x in range(1, 7) for y in range(1, 7)]
print(res)
```

```python
# return the dice combinations if the sum of two face value is divisible by 3
res = [(x, y) for x in range(1, 7) for y in range(1, 7) if (x+y)%3==0]
print(res)
```

```python
res = [(x, y) for x in range(1, 7) for y in range(1, 7) if x == 4 or y == 4]
print(res)
```

```python
# Count the number of vowels exists in the string
string = input()
res = [ char for char in string.lower() if char in "aeiou"]
print(res)
print(len(res))
```

```python
# Count the number of vowels exists in the string
string = input()
res = sum([1 for char in string.lower() if char in "aeiou"])
print(res)
```

```python
# Set comprehension 
# 1 to n numbers
n = int(input())
res = {x for x in range(1, n+1)}
print(type(res))
print(res)
```

```python
# Dictionary comprehension
# 1 to n numbers
n = int(input())
res = {x:x**2 for x in range(1,n+1)}
print(type(res))
print(res)
```

```python
# Tuple comprehension
# i to n numbers
n = int(input())
res = tuple(x for x in range(1, n+1))
print(type(res))
print(res)
```

```python
# Weather the person is eligible to vote or not
age = int(input())
if age >= 18:
    print("Eligible")
else:
    print("Not Eligible")
```

```python
# Assert Keyword : this keyword is used for testing.
age = int(input())
assert age >= 18, "Age must be above 17"
print("Eligible")
```
