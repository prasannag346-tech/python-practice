## sting methods
-case convention methods
      -upper() : It converts entire string into upper case.
      -lower() : It converts entire string into lower case.
      -title() : It converts the first letter of every eord into uppercase and remaining lowercase.
      -swapcase() : It converts upper to lower and lower to upper.
      -captalise() : It converts first letter into uppercase and remaining into lowercase.
      

```python
string="the sky is in the blue color"
print(string.upper())
print(string.lower())
print(string.title())
print(string.swapcase())
print(string.captalise())
```
