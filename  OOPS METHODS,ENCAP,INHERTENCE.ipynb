- Instance Method : The method defination takes first parameter as `self` those methods  calledas `Instance methods`. This methods works on instance attributes. <br>
  `syntax:`<br>
               `class class_name:`                  <br>
                  `  def method_name(self):`        <br>
                        ` #block code`
- Class Method : The method defination starts with `classmethod` decorator and it takes first parameter as `cls` those methods calles as `Class Methods`. this Methods works on class Attributes.
  `syntax:`<br>
      `        Class  class_name:`                    <br>
      `               @Classmethod`                    <br>
      `               def method_name(cls, parameters):`  <br>
      `                   #block code`         <br>
- Static Method : The method defination starts with `staticmethod` decorator and it doesn't take first parameter as `self or cls` those methods calles as `Stitic Methods`. This methods works on except instance attributes and cclass attributes.
- `Syntax:` <br>
      `         Class class_name:`                     <br>
      `               @staticmethod`                      <br>
      `               def method_name(parameters):`        <br>
      `                     #block code`    <br>
  
- Initilization constructor(__init__) : The instalization constructor is used to instilize the instants attribute data and this constructor takes first parameter as `self` and this constructor executes whenever an object is created for class.


```python
# Instance method
# class attribute - school
# instance attribute - id, name, class
class Student:
    school = "Codegnan School" # class attribute
    def __init__(self, stdid, name, standard):
        self.name = name  # instance attributes
        self.stdid = stdid # instance attributes
        self.standard = standard # instance attributes
    def student_details(self): # Instance Methods
        return f"I am {self.name} and i am from {self.standard} class"
    # class Methods
    @classmethod
    def update_school(cls, new_school_name):
        cls.school = new_school_name
        return "School name Updated"
    @staticmethod
    def percentage(marks:list):
        return(sum(marks)/300) * 100

        
s1 = Student(101, "Naveena", 10)
s2 = Student(102, "Srujana", 9)
print(s1.school, s2.school)
print(s1.student_details())
print(s1.update_school("CODEEDGE school"))
print(s1.school, s2.school)
print(s2.percentage([85,68,95]))
```

#### Encapsulation: 
     - Encapsulation means, wraping the methods and attributes in a single unit that one is class. The main aim of Encapsulation is to restrict the direct access of attributes and methods.
     - This type of restriction can implement by using `Access Modifiers` 
     - Types of Access Modifiers
         1. Public
         2. protected
         3. Private
     - In python we don't have any keywords to represent it public, protected and private, but by using underscore(_) can define.
         1. Public : The Public Members(i.e attributes + methods) doesn't starts with underscore(_). this public members can access anywhere in the program.
         2. Protected : The Protected Members(i.e attributes + methods) starts with single underscore(_). This members can assess in inside the class and In subclasses also.
         3. Private : The Private Members(i.e attributes + methods) starts with double underscore(__). This members can access in only inside.

- To acces the protected member ` obj._protected member name`
- To access the private members by using name handling
     `syntax:`
             `    obj._classesname__private number name   `    <br>
- The advantage of encapsulation is they provide the security for data asa well as for Functionalities.

```python
class Bank:
    def __init__(self, name, account, password, balance):
        self.name = name
        self.account = balance
        self.__password = password
        self._balance = balance
    def __total_users(self):  # getter method
        return "Total users : 101"
    def get_balance(self):  # getter method
        return f"Current Balance is:{self._balance}"
    def deposite(self, deposite_amount):  # setter method
        self._balance += deposite_amount
        return f" {deposite_amount} deposited Successfully"
        
    
u1 = Bank("prasanna", 1001, "prasanna@123", 10000)
print(u1.name)
print(u1.get_balance())
print(u1.deposite(5000))
print(u1.get_balance())
print(u1._balance)  # Protected members access in outside the class
print(u1._Bank__total_users())  # Name Mangling : syntax: obj._classname__attribute)
```

### Inheritance : In python, A class get properties from another class. This process is called as "Inheritance".
- A child class is inherited from the parent class.
  `      syntax:`                            <br>
  `             class parent_class_name:`   <br>
  `                       #attributes`<br>
  `                       #methods`               <br>
  `             class child class_name(parent_class_name)`      <br>
  `                       #attributes`   <br> 
  `                       #methods`      <br>
   
## Types of Inheritance
    1. Single Inheritance
    2. Multi level Inheritance
    3. Mutliple Inheritance
    4. Hierarchial Inheritance
    5. hybrid Inheritance
## Single Inheritance : A child class get properties from single parent class. This is called as "Single Inheritance".
        child class ----------> parent class
         crow  ------------------> Birds
         apple -------------------> Fruits
## Multi level Inheritance : A class (i.e child class) get properties from another class (i.e parent class), that class get properties from some another class(i.e grand parents).
            child class -------> parent class --------> grand parent class
              puppy  ------------> dog --------------------> Animal
              child ----------------> father ---------------> grand father
## Multiple Inheritance : A Single child class get properties from the muiltple parent classes. This is called as "Multiple inheritance".
## Hierarchial Inheritance : A Multiple child classes grt properties from the single parent class, This is called as " Hirearchial Inheritance".
        parent -> c1 ----> c2 ---------> c3
         milk --> curd --> paneer ---> buttermilk
        university --> CSE ---> CSM --> MECH
        Universe --> sun --> moon --> stars
## Hybrid Inheritence : This is a combination of two or more types of inheritences.

```python
# Multi Level inheritance
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"My self {self.name} and i am {self.age} year old"
    class Student(Person):
    def __init__(self, name, age, branch):
        super().__init__(name, age)
        self.branch = branch
        
    class Employee(Student):
    def __init__(self, name, age, branch, salary):
        super().__init__(name, age, branch)
        self.salary = salary
    def get_salary(self, bonus=0):
        self.salary += bonus
        return self.salary
   
  
e = Employee("prasanna", 21, "CSE", 50000)
print(e.name)
print(e.branch)
print(e.details())
print(e.get_salary())
print(e.get_salary(bonus = 10000))
```
