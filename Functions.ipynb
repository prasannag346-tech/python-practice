```python
# 3. With parameters and With return
#by default functions None value, when function not returning any value
# Function defination block
def addition(x:int, y:int)->None:
    c = x + y
    print(f"This is from Function defination block:{c}")
    
# function calling block
a, b = 10,20
res = addition(a, b)
print(f"This is from Function calling block:{res}")
```

```python
# 2. Without parameters and with return
# Function defination block
def addition()->int:
    x = 10
    y = 20
    z = x + y
    return z
    
# function calling block
res = addition()
print(f"This is from Function calling block:{res}")
```

```python
# implement even or odd using functions
#function defination block
def even_or_odd(x:int)->str:
    if x % 2 == 0:
        return "Even"
    else:
        return "odd"
#function calling block
n = int(input())
print(even_or_odd(n))

```

```python
# implement prime number program using functions
# function defination block
def check_prime(x:int)->str:
        if x < 2:
            return "Not prime"
        for i in range(2, x//2 + 1):
            if x % i == 0:
                return "Not prime"
        return "Prime"
# function calling block
num = int(input())
print(check_prime(num))
        
```
