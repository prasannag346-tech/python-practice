```python
# List
my_list = ['a', 'b', 'c']
# List to String
string = ''.join(my_list)
print("List to String:", string)
# List to Tuple
tup = tuple(my_list)
print("List to Tuple:", tup)
# List to Set
st = set(my_list)
print("List to Set:", st)
```

```python
# Tuple
my_tuple = ('p', 'y', 't', 'h', 'o', 'n')
# Tuple to String
string = ''.join(my_tuple)
print("Tuple to String:", string)
# Tuple to List
lst = list(my_tuple)
print("Tuple to List:", lst)
# Tuple to Set
st = set(my_tuple)
print("Tuple to Set:", st)
```

```python
# list to dictionary conversion
l = [('a',1), ('b',2)]
print(dict[l])
```
