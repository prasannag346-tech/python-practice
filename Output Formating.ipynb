### Output Formating

```python
name = 'srinu'
age = 20
print("my name is srinu and i'm 20 years old")
```

```python
# 1. string formating using comma
name = 'srinu'
age = 20
# my name is srinu and i am 20 years old
print("My name is", name,"and i am  ",age,"years old") #using comma
print("My name is "+name+" and i am "+str(age)+" years old") #string using concatanation
print(f"My name is {name} and i am {age} years old")    #F format
print("My name is {} and i am {} years old".format(name,age))   #Dot Format
print("My name is %s and i am %d years old"%(name,age))
```

```python
# Access Specifiers
# string - %s
# integer - %d
# float - %f

age=21
name="prasanna"
print("My Name is %s and i am %d years old."%(name,age))
print("My Name is %s and i am %d years old."%(age,name))
```
