### Set Charactristics

```python
# Set unordered collection
#set returns unique elements
s = {1,2,3,2,3,'hi','a',2.02, (10,20,30)}
print(s)
print(s[0])  set does not support indexing
```

```python
# List can't be used as set element
s = {1,2,[3,2,3],'hi','a',2.02, (10,20,30)}
print(s)
```

### Set Methods

```python
# Element adding method:
# add(ele): It adds the elements to set if element is not in set,otherwise nothing
s={1,2,3,4,5,4,3}
print(s)
s.add(100)
s.add('hi')
print(s)
```

```python
# sets operation
# union(set): It returns the all unique elements from all the sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {1,2,8,9}
union = s1.union(s2,s3)
print(union)
res = s1|s2|s3
print(res)
```

```python
s1 = {1,3,5,7}
s2 = {4,5,6,7}
intersection = s1.intersection(s1,s2)
print(intersection)
res = s1 & s2
print(res)
```

```python
# sets operation
# intersection(set): It returns the common elements from all the sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {5,1,2,8,9}
intersection = s1.intersection(s2,s3)
print(intersection)
res = s1 &s2 & s3
print(res)
```

```python
# sets operation
# difference(set):It returns the a set elements those are not in another sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {5,1,2,8,9}
intersection = s1.difference(s2,s3)
print(intersection)
res = s1 - s2 - s3
print(res)
print((s1 | s2) - s3)
```

```python
#set operations
# symmetric_difference(set) : It return the non common elements from all the sets
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {5,1,2,8,9}
sy_diff = s1.symmetric_difference(s2)
print(sy_diff)
```

```python
# issubset(set)
s1 = {1,3,5,7}
s2 = {6,5,7}
subset=s1.issubset(s2)
print(subset)
```

```python
# issubset(set)
s1 = {1,2,3}
s2 = {1,2,3,4}
print(s1.issubset(s2))
print(s2.issubset(s1))
```

```python
# issuperset(set)
s1 = {1,2,3}
s2 = {1,2,3,4}
print(s2.issuperset(s1))
```

```python
# isdisjointset(set)
s1 = {1,2,3}
s2 = {1,2,3,4}
print(s1.isdisjoint(s2))
```

```python
# intersection_update(set)
s1 = {1,3,5,7}
s2 = {4,5,6,7}
s3 = {5,1,2,8,9}
s1.intersection_update(s1,s2,s3)
print(s1,s2,s3)
```
