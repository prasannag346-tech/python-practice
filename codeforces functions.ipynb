```python
# print
def printNumbers(n):
    for i in range(1, n+1):
        print(i, end = " ")
n = int(input())
printNumbers(n)
```

```python
# Functions sheet
# add
def add(x, y):
    return x+y
n, m = map(int,input().split())
print(add(n, m))
```

```python
# Find the binary value of a number
n = int(input())
res = ""
while n>0:
    rem = n%2
    res = str(rem) + res
    n = n // 2
print(res)
```

```python
# reverse a string without using slicing
string = input()
rev = ""
for char in string:
    rev = char + rev
print(rev)
```

```python
# find the binary value and check wheather the binary value id palindrome or notn = int(input())
n = int(input())
res = ""
while n>0:
    rem = n%2
    res = str(rem) + res
    n = n // 2
print(res)
rev = ""
for char in res:
    rev = char + rev
print(rev)
print("palindrome" if res == rev else "Not Palindrome")
```
