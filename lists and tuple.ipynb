### List Methods   
- Item adding methods
  - append(item):It adds items at the end of the list
  - extend(iterator):It adds the all iterator elements at the end of the list
  - insert(pos,item):It adds items at specific position

```python
list=[10,20,30,40]
list.append(100)
print(list)
list.extend([1,2,3]) 
print(list)
list.append([100,200])
print(list)
list.insert(0,1000)
print(list)
```

### Item Removing Methods
 - remove(item):It removes the first occurance item from the list if it is exists in the list otherwise 'ValueError'
 - pop(pos):It removes and return the specified position item if position is passed,otherwise it remove and returns the last item from the list
 - clear():It removes all elements from the list

```python
list = [10,20,30,40]
list.remove(20)
print(list)
```

```python
list = [10,20,30,40]
list.remove(20)
print(list)
list.remove(100)
```

```python
list = [10,20,30,40,50,20]
val = list.pop()
print(val)
print(list)
val2 = list.pop(0) #It removes the 0th position
print(val2)
print(list)
list.clear
print(list)
```

```python
val2 = list.pop(1)
print(val2)
print(list)
list.clear
print(list)
```
