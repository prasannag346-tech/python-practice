### File Handling :
- File : The file is used to store the dats in electronic format in the system.  <br>
`   File Operations: `    <br>
     1. Opening a file       - open()   <br>
     2. Reading a file       - read()    <br>
     3. writing into a file  - write()    <br>
     4. closing a file       - close()   <br>
### Opening a file:
- To open a file by using open() in that, we need to pass file name and file opening mode.   <br>
`        syntax: `   <br>
               file_obj = open(file_name, mode)  <br>
                 # read / write operations  <br>
                 # close file  <br>
### File Opening Names:
- r - (read) - opening file in read mode and if file not exist it raises FILENOTFOUNDError.
- w - (write) - Opening a file in write mode if file not exist it creates newfile otherwise it erases the previous content.
- a - (append) - Opening a file in append mode, if file not exists it creates a newfile otherwise it add content to the previous content.
- x - (create) - Create New  file .
- b - (binary) - This mode is used to perform operation on binary files. For writing (wb) mode and for reading (rb) mode.
- r+- (read + write) 
- w+- (write + read) 
- a+- (append + read)


### Reading content from a file : 
- To perform the read operation on the file, for that we need to open file in 'r' mode, and then we follow the operations.
- read() : It returns entire file content as single string.
- readline() : This method returns first line.
- readlines() : It returns the list of lines.

### Writing Into a file : 
- If you open a file in write mode we can perform the following operations. 
- In this we have 2 methods:
1. write(string) : It add string into a file.
2. writelines(List-of-lines) : It adds all list lines to a file.

### Closing a file : 
- After performing the read/write operations on a file, we need to close the file by using close(), otherwise data loss will happen.
