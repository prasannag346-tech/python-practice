```python
# return 1 to n numbers using functions
def fun(n):
    res = []
    for i in range(1, n+1):
        res.append(i)
    return res
print(fun(10))
```

```python
# return 1 to n numbers using functions
import sys
def fun(n):
    lst = []
    for i in range(1, n+1):
        lst.append(i)
    return lst
r = fun(10)
print(sys.getsizeof(r))
print(r)
```

```python
# return 1 to n numbers using generators
import sys
def gen(n):
    lst = []
    for i in range(1, n+1):
        yield i
r = gen(3)
print(r)
n = next(r)
print(sys.getsizeof(n))
print(n)
print(next(r))
print(next(r))
print(next(r))
print(next(r))  #stopiteration,Because no element left to get value in generator object.
```

```python
import time
import sys
s = time.time()
def gen(n):
    for i in range (1, n+1):
        yield i ** 3
r = gen(100)
print(r)
#to prevent the stopIteration error by using for loop
for val in r:
    print(val, end=" ")
e = time.time()
print(e-s)
```

```python
def gen(x):
    x = x + 1
    yield x
    x = x + 2
    x = x + 3
    yield x
    x += 5
    x += 6
    yield x
res = gen(10)
print(res)
print(next(res))
print(next(res))
```
