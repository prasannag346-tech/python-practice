```python
import re
```

```python
# match
string = "The sky is blue and ocean is blue"
res = re.match("The", string)
print(res)
print(res.span()) # it returns the matched pattern positions
print(res.group()) # it returns the Matched result
```

```python
# match
string = "The sky is blue and ocean is blue"
res = re.match("sky", string)
print(res)
print(res.span()) # it returns the matched pattern positions
print(res.group()) # it returns the Matched result
```

```python
# match(pat, string) : It checks the weather the string is starts with pattern or not
try:
    string = "The sky is blue and ocean is blue"
    res = re.match("sky", string)
    print(res)
    print(res.span()) # it returns the matched pattern positions
    print(res.group()) # it returns the Matched result
except:
    print("There is no matched pattern")
```

```python
# search(pat, string) : It checks the entire string and it return the first matched pattern
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("sky", string)
    print(res)
    print(res.span()) # it returns the matched pattern positions
    print(res.group()) # it returns the Matched result
except:
    print("There is no matched pattern")
```

```python
# search(pat, string) : It checks the entire string and it return the first matched pattern
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("z", string)
    print(res)
    print(res.span()) # it returns the matched pattern positions
    print(res.group()) # it returns the Matched result
except:
    print("There is no matched pattern")
```

```python
# search(pat, string) : It checks the entire string and it return the first matched pattern
try:
    string = "The sky is blue and ocean is blue"
    res = re.search("blue", string)
    print(res)
    print(res.span()) # it returns the matched pattern positions
    print(res.group()) # it returns the Matched result
except:
    print("There is no matched pattern")
```

```python
# findall(pat, string) : It returns the all matched patterns as list
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("blue", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall(pat, string) : It returns the all matched patterns as list
try:
    string = "The sky is blue and ocean is blue"
    res = re.findall("[abcl]", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# split(pat, string) : It splits the string wherever the pattern is matched.
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("blue", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# split(pat, string) : It splits the string wherever the pattern is matched.
try:
    string = "The sky is blue and ocean is blue"
    res = re.split("[abclo]", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# sub(old_pat,new_pat, string) : It changes the old string to the new string.
try:
    string = "The sky is blue and ocean is blue"
    res = re.sub("blue","red", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall(pat, string) : It returns the all matched patterns as list
try:
    string = "The sky is blue and ocean is bl65ue"
    res = re.findall("[a-zA-Z]+", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# findall(pat, string) : It returns the all matched patterns as list
try:
    string = "The sky is blue 33 and ocean is bl65ue 1234"
    res = re.findall(r"[\w]+", string)
    print(res)
except:
    print("There is no matched pattern")
```

```python
# Validate mobile number
try:
    mobile = input()
    res = re.fullmatch("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f" {mobile} is valid")
    else:
        print(f" {mobile} is Invalid")
except: 
    print("There is no matched pattern")
```

```python
# PAN number Validation
try:
    pan_number = input()
    res = re.fullmatch("^[A-Z]{5}[0-9]{4}[A-Z]$", pan_number)
    print(res)
    if res:
        print(f" {pan_number} is valid")
    else:
        print(f" {pan_number} is Invalid")
except:
    print("There is no matched pattern")
```

```python
# Aadhar number validation
try:
    aadhar_number = input()
    res = re.fullmatch(r"^[2-9][0-9]{3}\s?[0-9]{4}\s?[0-9]{4}", aadhar_number)
    print(res)
    if res:
        print(f" {aadhar_number} is valid")
    else:
        print(f" {aadhar_number} is Invalid")
except:
    print("There is no matched pattern")
```
