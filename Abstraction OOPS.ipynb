### Abstraction 

```python
from abc import ABC,abstractmethod
class Mobile(ABC):

    @abstractmethod
    def ram(self):
        pass
    @abstractmethod
    def rom(self):
        pass
    @abstractmethod
    def battery_size(self):
        pass
    # concreate method implementation
    def display_quality(self, Hzs):
        return f"This mobile display quality is {Hzs} Hzs"
class Oppo(Mobile):
    def __init__(self, model, battery_size, display_quality):
        self.model = model
        self.battery = battery_size
        self.display = display_quality
    def ram(self, ram_size):
        return f" The Oppo mobile {self.model} has {ram_size} GB RAM"
    def rom(self, rom_size):
        return f" The Oppo Mobile {self.model} has {rom_size} GB ROM"
    def battery_size(self):
        return f" The Oppo Mobile{self.model} has {self.battery} mah Battery capaciyty"
    def camera(self, primary, secondary):
        return f"The Oppo {self.model} mobile has a {primary}MP primary camera and a {secondary}MP secondary camera."
        
# m = Mobile() Typeerror: Can't instantiate abstract class Mobile without an implemention for abstract methods 'battery', 'ram', 'rom'
reno15 = Oppo("reno15", 5000, 120)
print(reno15.ram(4))
print(reno15.rom(128))
print(reno15.battery_size())
print(reno15.camera(32, 52))
print(reno15.display_quality(120))

 
```

```python
# Find the area of perimeter 2d
from abc import ABC, abstractmethod

class Shape(ABC):

    @abstractmethod
    def calculate(self):
        pass


class Rectangle(Shape):
    def _init_(self, length, width):
        self.length = length
        self.width = width

    def calculate(self):
        area = self.length * self.width
        perimeter = 2 * (self.length + self.width)

        print("Area =", area)
        print("Perimeter =", perimeter)

r = Rectangle(10, 5)
r.calculate()
```
