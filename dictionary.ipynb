### Dictionary Methods

```python
 - get(key, default_value:Optional):It returns the value if the key is exists in dictionary,otherwise it returns default value.
```

```python
# dictionary keys are unique 
details = {"age":25,"name":"babu","course":"python","batch":57,"age":21,"location":"Hyderabad"}
print(details['name'])
# print(details['marks']) # This one raises keyerror
print(details.get('name'))
print(details.get('marks')) # get method bydefault it returns none if key does not exists in dictionary
print(details.get('marks',"marks not available")) #it returns default value because , key not exists 
 
```

```python
- update(new_dictionary) : This method updates the value if new_dictionary keys are already exists in dictionary,otherwise it adds new items to dictionary.
```

```python
details = {"age":25,"name":"babu","course":"python","batch":57}
details.update({'batch':58,'course':'Java','location':'Hyderabad'})
print(details)
```

```python
- keys():This method returns the list dictionary keys
- value():This method returns the list dictionary values
- itemsm():This method returns the list of tuples of key-value pairs
```

```python
details = {"age":25,"name":"babu","course":"python","batch":57,"age":21,"location":"Hyderabad"}
print(details.keys())
print(details.values())
print(details.items())
```

### Items removing methods
- popitems():It returns and removes the last item as tuple
- pop(key):this removes the specified key item and returns the specified key value,otherwise keyerror

```python
details = {"age":25,"name":"babu","course":"python","batch":57,"age":21,"location":"Hyderabad"}
item = details.popitem()
print(item)
print(details)
val = details.pop('name')
print(val)
print(details)
```
